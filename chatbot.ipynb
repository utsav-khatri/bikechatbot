```python
import json
import random

with open("bike.json", "r", encoding="utf-8") as file:
    data = json.load(file)

off_topic_reply = [
    "I am a Bike Information Chatbot only.",
    "Please ask questions related to motorcycles.",
    "Try asking about KTM, Yamaha, BMW, Kawasaki, Ducati, Honda, Suzuki etc.",
    "Example:\n• Tell me about Hayabusa\n• Price of BMW S1000RR\n• Mileage of Yamaha R15 V4"
]

# Keywords -> JSON field names
parameters = {
    "engine": "engine_cc",
    "power": "power_hp",
    "horsepower": "power_hp",
    "torque": "torque_nm",
    "mileage": "mileage_kmpl",
    "price": "price_inr",
    "top speed": "top_speed_kmph",
    "speed": "top_speed_kmph",
    "fuel tank": "fuel_tank_l",
    "weight": "weight_kg",
    "seat height": "seat_height_mm",
    "ground clearance": "ground_clearance_mm",
    "wheelbase": "wheelbase_mm",
    "transmission": "transmission",
    "cooling": "cooling",
    "abs": "abs",
    "traction": "traction_control",
    "quick shifter": "quick_shifter",
    "ride modes": "ride_modes",
    "headlight": "headlight",
    "console": "console",
    "bluetooth": "bluetooth",
    "navigation": "navigation",
    "launch year": "launch_year",
    "emission": "emission",
    "rating": "rating",
    "country": "country",
    "color": "colors",
    "colour": "colors",
    "service": "service_interval_km",
    "warranty": "warranty_years"
}


def find_bike(text):
    for bike in data["bikes"]:
        if bike["name"].lower() in text:
            return bike
    return None


def show_all_details(bike):
    print("\n🏍️", bike["name"])
    print("-" * 50)

    for key, value in bike.items():
        if key == "name":
            continue

        if isinstance(value, list):
            value = ", ".join(map(str, value))

        print(f"{key.replace('_',' ').title()} : {value}")


def show_parameter(bike, user_input):
    for keyword, field in parameters.items():

        if keyword in user_input:

            value = bike.get(field, "Not Available")

            if isinstance(value, list):
                value = ", ".join(map(str, value))

            print("\n🏍️", bike["name"])
            print("-" * 40)
            print(f"{keyword.title()} : {value}")
            return True

    return False


print("=" * 60)
print("🏍️ Welcome to BikeBot")
print("=" * 60)
print("I answer ONLY bike related questions.")
print("Examples:")
print("• Tell me about Yamaha R15 V4")
print("• Engine of KTM RC 390")
print("• Mileage of Hayabusa")
print("• Price of BMW S1000RR")
print()

while True:

    user = input("You : ").lower().strip()

    if user == "exit":
        print("BikeBot : Ride Safe! 🏍️")
        break

    bike = find_bike(user)

    if bike:

        # Show only requested parameter if present
        if not show_parameter(bike, user):
            # Otherwise show complete details
            show_all_details(bike)

    else:
        print(random.choice(off_topic_reply))
```
